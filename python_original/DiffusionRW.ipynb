### Simulation of a diffusive process in 1D

```python
import numpy as np
import matplotlib.pyplot as plt

N = 100 # number of particles
S = 100 # number of steps
rm = np.random.randn(N, S) # generates a NxS matrix of random numbers
srm = np.sign(rm) # takes the sign and assigns +1 or -1
trayect = np.cumsum(srm, axis=1) # calculates trajectory
trayect = np.concatenate((np.zeros((N, 1)), trayect), axis=1) # start at x=0
S = S + 1

# Plot the trajectories of all particles
for k in range(N):
    plt.plot(trayect[k, :], np.arange(S), '-o', markersize=3)
plt.gca().tick_params(labelsize=14)
plt.ylabel('number of steps (t/$\\tau$)')
plt.xlabel('displacement (x/$\\delta$)')
plt.title(f"Diffusion in 1D: {N} particles, {S-1} steps", fontsize=14)
plt.show()
```

### Simulation of a diffusive process in 2D

```python
import numpy as np
import matplotlib.pyplot as plt

N = 100 # number of particles
S = 100 # number of steps
rm_x = np.random.randn(N, S)
srm_x = np.sign(rm_x)
trayect_x = np.cumsum(srm_x, axis=1)

rm_y = np.random.randn(N, S)
srm_y = np.sign(rm_y)
trayect_y = np.cumsum(srm_y, axis=1)

trayect_x[:, 0] = np.zeros(N)
trayect_y[:, 0] = np.zeros(N)

plt.plot(trayect_x.T, trayect_y.T)
plt.xlabel(r'$\delta x$')
plt.ylabel(r'$\delta y$')
plt.title(f"Diffusion in 2D: {N} particles ; {S} steps", fontsize=14)
plt.show()
```

### Simulation of a diffusive process in 3D

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

N = 10 # number of particles
S = 100 # number of steps

rm_x = np.random.randn(N, S)
srm_x = np.sign(rm_x)
trayect_x = np.cumsum(srm_x, axis=1)

rm_y = np.random.randn(N, S)
srm_y = np.sign(rm_y)
trayect_y = np.cumsum(srm_y, axis=1)

rm_z = np.random.randn(N, S)
srm_z = np.sign(rm_z)
trayect_z = np.cumsum(srm_z, axis=1)

fig = plt.figure()
ax = plt.axes(projection='3d')
ax.scatter3D(trayect_x.T, trayect_y.T, trayect_z.T, c=trayect_x, cmap='Greens')
ax.view_init(10, 30)
ax.set_xlabel(r'$\delta x$')
ax.set_ylabel(r'$\delta y$')
ax.set_zlabel(r'$\delta z$')
ax.set_title(f"Diffusion in 3D: {N} particles ; {S} steps", fontsize=14)
plt.show()
```
